```python
import csv
import pandas as pd
```

```python
print(pd.read_csv("pengiriman.csv"))
```

```python
def hitung_total(daftar_angka):
    total = 0
    for angka in daftar_angka:
        total = total + angka
    return total

def hitung_rata_rata(daftar_angka):
    return hitung_total(daftar_angka) / len(daftar_angka)

def baca_ongkir(nama_file):
    ongkir = []
    with open(nama_file, newline="", encoding="utf-8") as file_csv:
        pembaca = csv.DictReader(file_csv)
        for baris in pembaca:
            ongkir.append(int(baris["ongkir"]))
    return ongkir

def hitung_terlambat(nama_file):
    jumlah = 0
    with open(nama_file, newline="", encoding="utf-8") as file_csv:
        pembaca = csv.DictReader(file_csv)
        for baris in pembaca:
            if baris["status"] == "terlambat":
                jumlah = jumlah  + 1
    return jumlah
```

```python
data = baca_ongkir("pengiriman.csv")
print(data)

print("Jumlah paket:", len(data))
print("Total ongkir:", hitung_total(data))
print("Rata-rata ongkir:", hitung_rata_rata(data))
print("Ongkir termahal:", max(data))
print("Ongkir termurah:", min(data))
print("Jumlah paket terlambat", hitung_terlambat("pengiriman.csv"))
```
