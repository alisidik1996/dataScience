```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("pengiriman.csv")
```

```python
df["tanggal"] = pd.to_datetime(df["tanggal"], format="mixed", dayfirst="True")
df["kota"] = df["kota"].astype("category")
df["status"] = df["status"].astype("category")
df["ongkir"] = df["ongkir"].fillna(df["ongkir"].median())
```

```python
df.groupby("kota", observed="True")["ongkir"].sum()
```

```python
df["status"].value_counts()
```

```python
df.groupby("status", observed="True")["ongkir"].mean()
```

```python
df.groupby("kota", observed="True")["ongkir"].sum().plot(kind="bar")
plt.title("Total Ongkir per Kota")
plt.ylabel("Ongkir (Rp)")
plt.tight_layout()
plt.show()
```

```python
data = [
    {"Kota":"Jakarta", "ongkir":45000, "terlambat":False},
    {"Kota":"Bandung", "ongkir":5000, "terlambat":True},
    {"Kota":"Surabaya", "ongkir":15000, "terlambat":False},
    {"Kota":"Medan", "ongkir":40000, "terlambat":True},
    {"Kota":"Tasik", "ongkir":25000, "terlambat":False}
]
```

```python
total = 0
for p in data:
    total = total + p["ongkir"]
print("Total ongkir:", total)
```

```python
rata = total / len(data)
print("Rata-rata ongkir:", rata)
```
