```python
import pandas as pd
import io
```

```python
csv_data = """tanggal,kota,berat_kg,ongkir,status
2026-11-10,Jakarta,12,45000,terkirim
2026-11-11,Bandung,8,32000,terkirim
2026-11-12,Surabaya,20,78000,terlambat
2026-11-13,Jakarta,15,52000,terkirim
2026-11-14,Bandung,5,25000,terkirim
2026-11-15,Yogyakarta,3,60000,terkirim
"""
```

```python
df = pd.read_csv(io.StringIO(csv_data))
print(df)
```

```python
print(df.groupby("status")["ongkir"].mean())
print("Rata-rata berat:", df["berat_kg"].mean())
print("Total ongkir:", df["ongkir"].sum())
print("Ongkir Termahal", df["ongkir"].max())
print("Pengiriman Terlambat:", df.groupby("status")["status"].count())
```
