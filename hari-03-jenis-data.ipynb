```python
import pandas as pd
import io
```

```python
csv_data = """tanggal,kota,berat_kg,ongkir,status
2026-11-10,Jakarta,12,45000,terkirim
2026-11-11,Bandung,8,32000,terkirim
2026-11-12,Surabaya,20,78000,tertelat
2026-11-13,Jakarta,5,,terkirim
13/11/2026,Bandung,15,60000,terkirim
2026-11-14,Jakarta,25,95000,tertelat
2026-11-15,Surabaya,10,41000,terkirim
"""
```

```python
df = pd.read_csv(io.StringIO(csv_data))
```

```python
df.dtypes
```

```python
df.isna().sum()
```

```python
# mengubah format tanggal dari 'str' menjadi 'dateTime'
df["tanggal"] = pd.to_datetime(df["tanggal"], format="mixed", dayfirst="True")

# merubah 'kota' dan 'status' menjadi tipe 'category' sudah jadi label
df["kota"] = df["kota"].astype("category")
df["status"] = df["status"].astype("category")
```

```python
# mengisi kolom na di kolom ongkir dengan nilai tengah
df["ongkir"] = df["ongkir"].fillna(df["ongkir"].median())
```

```python
print(df)
df.describe(include='all')
```

```python
# menghitung rata" berat pengiriman tiap kota
df.groupby("kota", observed=True)["berat_kg"].mean()
```
